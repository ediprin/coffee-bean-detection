```python
from google.colab import drive
drive.mount('/content/drive')
import importlib, os, shutil, subprocess, sys
from pathlib import Path
REPO=Path('/content/coffee-bean-detection')
BRANCH='codex/coffee-standard-primary-audit'
REMOTE='https://'+'github.com/ediprin/coffee-bean-detection.git'
os.chdir('/content')
if REPO.exists(): shutil.rmtree(REPO)
subprocess.run(['git','clone','--depth','1','--branch',BRANCH,REMOTE,str(REPO)],check=True)
subprocess.run([sys.executable,'-m','pip','install','-q','-e',str(REPO),'gdown'],check=True)
sys.path.insert(0,str(REPO/'src')); importlib.invalidate_caches(); os.chdir(REPO)
from coffee_detector.analysis.coffee_standard_j25_thesis_provenance import audit_j25_thesis_provenance
from coffee_detector.drive_project import resolve_drive_project_root
PROJECT=resolve_drive_project_root()
ARCHIVE=Path('/content/data_aug_11.zip')
if not ARCHIVE.is_file():
    url='https://drive.google.com/uc?id=1AofT7VbiNFM8ul-0vyCAKj7Rp4j5OX0f'
    subprocess.run([sys.executable,'-m','gdown',url,'-O',str(ARCHIVE)],check=True)
OUTPUT=PROJECT/'evidence/coffee-standard-j25-thesis-provenance-v1/coffee_standard_j25_thesis_provenance.json'
print('REPO:',subprocess.check_output(['git','rev-parse','--short','HEAD'],text=True).strip())
print('ARCHIVE:',ARCHIVE,ARCHIVE.stat().st_size)
```
