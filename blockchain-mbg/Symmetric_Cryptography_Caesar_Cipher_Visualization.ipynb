# Enkripsi Simetris Sederhana dengan Caesar Cipher

Notebook ini memperagakan: **Plaintext + Key → Encryption → Ciphertext → Decryption → Plaintext**.

Caesar Cipher dipakai karena sangat sederhana dan mudah divisualisasikan. Ini hanya untuk pembelajaran, bukan keamanan nyata.

```python
import string
import pandas as pd
import matplotlib.pyplot as plt
print('Library siap digunakan.')
```

## 1. Fungsi Enkripsi dan Dekripsi

Jika key = 3, maka A → D, B → E, C → F.

```python
def caesar_encrypt(plaintext, key):
    ciphertext = ''
    for char in plaintext:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            ciphertext += chr((ord(char) - base + key) % 26 + base)
        else:
            ciphertext += char
    return ciphertext

def caesar_decrypt(ciphertext, key):
    plaintext = ''
    for char in ciphertext:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            plaintext += chr((ord(char) - base - key) % 26 + base)
        else:
            plaintext += char
    return plaintext
```

## 2. Masukkan Plaintext dan Key

Contoh: plaintext `HALO DUNIA`, key `3`.

```python
plaintext = input('Masukkan plaintext: ')
key = int(input('Masukkan key (bilangan bulat): ')) % 26

ciphertext = caesar_encrypt(plaintext, key)
decrypted_text = caesar_decrypt(ciphertext, key)

print('\nPlaintext      :', plaintext)
print('Key            :', key)
print('Ciphertext     :', ciphertext)
print('Hasil dekripsi :', decrypted_text)
```

## 3. Visualisasi Alur Enkripsi Simetris

```python
fig, ax = plt.subplots(figsize=(12, 3))
ax.axis('off')
positions = [0.08, 0.32, 0.56, 0.80]
labels = [
    f'Plaintext\n{plaintext}',
    f'Encryption\nShift +{key}',
    f'Ciphertext\n{ciphertext}',
    f'Decryption\nShift -{key}\n{decrypted_text}'
]
for x, label in zip(positions, labels):
    ax.text(x, 0.5, label, ha='center', va='center', bbox=dict(boxstyle='round,pad=0.5'))
for i in range(len(positions)-1):
    ax.annotate('', xy=(positions[i+1]-0.08, 0.5), xytext=(positions[i]+0.08, 0.5), arrowprops=dict(arrowstyle='->'))
ax.set_title('Visualisasi Enkripsi dan Dekripsi Caesar Cipher')
plt.show()
```

## 4. Visualisasi Pergeseran Alfabet

```python
alphabet = string.ascii_uppercase
shifted = alphabet[key:] + alphabet[:key]
mapping_df = pd.DataFrame({
    'Plain Alphabet': list(alphabet),
    'Cipher Alphabet': list(shifted)
})
display(mapping_df)
```

## 5. Visualisasi Karakter Plaintext → Ciphertext

```python
rows = []
for p, c in zip(plaintext, ciphertext):
    rows.append({'Plaintext': p, 'Ciphertext': c, 'Berubah?': 'Ya' if p != c else 'Tidak'})
char_df = pd.DataFrame(rows)
display(char_df)
```

## 6. Uji Key yang Salah

```python
wrong_key = (key + 1) % 26
wrong_result = caesar_decrypt(ciphertext, wrong_key)
print('Ciphertext         :', ciphertext)
print('Key benar          :', key)
print('Dekripsi key benar :', decrypted_text)
print('\nKey salah          :', wrong_key)
print('Dekripsi key salah :', wrong_result)
```

## Kesimpulan

`C = E(P, K)` dan `P = D(C, K)`, dengan P = plaintext, C = ciphertext, dan K = secret key.
