# Blockchain Sederhana untuk Transaksi Penjualan Hasil Pertanian

Alur:
**Data transaksi pertanian → SHA-256 → hash transaksi → Genesis Block → Block 1 → uji perubahan data**

Tujuan: memahami integritas data transaksi menggunakan SHA-256 pada struktur blockchain sederhana.


```python
import hashlib
import json

def sha256(data: str) -> str:
    return hashlib.sha256(data.encode('utf-8')).hexdigest()

def hash_transaction(transaction: dict) -> str:
    transaction_string = json.dumps(
        transaction,
        sort_keys=True,
        separators=(',', ':')
    )
    return sha256(transaction_string)

def hash_block(block: dict) -> str:
    block_string = json.dumps(
        block,
        sort_keys=True,
        separators=(',', ':')
    )
    return sha256(block_string)

print('Fungsi hashing siap digunakan.')
```

## 1. Data Transaksi Penjualan Hasil Pertanian

Contoh transaksi dari petani ke koperasi.


```python
tx1 = {
    'transaction_id': 'AGR001',
    'farmer': 'Petani-A',
    'buyer': 'Koperasi-Tani',
    'commodity': 'Jagung',
    'quantity': 500,
    'unit': 'kg',
    'price_per_kg': 6500,
    'timestamp': '2026-09-10T08:00:00'
}

tx2 = {
    'transaction_id': 'AGR002',
    'farmer': 'Petani-B',
    'buyer': 'Koperasi-Tani',
    'commodity': 'Beras',
    'quantity': 300,
    'unit': 'kg',
    'price_per_kg': 14000,
    'timestamp': '2026-09-10T09:00:00'
}

tx3 = {
    'transaction_id': 'AGR003',
    'farmer': 'Petani-C',
    'buyer': 'Koperasi-Tani',
    'commodity': 'Kopi',
    'quantity': 120,
    'unit': 'kg',
    'price_per_kg': 85000,
    'timestamp': '2026-09-10T10:00:00'
}

transactions = [tx1, tx2, tx3]

for tx in transactions:
    total = tx['quantity'] * tx['price_per_kg']
    print(tx)
    print('Total transaksi: Rp{:,.0f}'.format(total).replace(',', '.'))
    print('-' * 80)
```

## 2. Menghitung Hash SHA-256 Setiap Transaksi


```python
transaction_hashes = []

for tx in transactions:
    tx_hash = hash_transaction(tx)
    transaction_hashes.append(tx_hash)
    print('ID   :', tx['transaction_id'])
    print('Hash :', tx_hash)
    print('-' * 80)
```

## 3. Membuat Genesis Block


```python
genesis_block = {
    'index': 0,
    'previous_hash': '0',
    'timestamp': '2026-09-10T00:00:00',
    'transactions': []
}

genesis_hash = hash_block(genesis_block)

print('Genesis Block:')
print(json.dumps(genesis_block, indent=2))
print('\nGenesis Hash:')
print(genesis_hash)
```

## 4. Membuat Block 1


```python
block1 = {
    'index': 1,
    'previous_hash': genesis_hash,
    'timestamp': '2026-09-10T12:00:00',
    'transaction_hashes': transaction_hashes
}

block1_hash = hash_block(block1)

print('Block 1:')
print(json.dumps(block1, indent=2))
print('\nPrevious Hash :', block1['previous_hash'])
print('Block 1 Hash  :', block1_hash)
```

## 5. Uji Perubahan Data

Harga jagung diubah dari **Rp6.500/kg** menjadi **Rp5.500/kg** untuk melihat perubahan hash.


```python
tx_uji = tx1.copy()

hash_sebelum = hash_transaction(tx_uji)
print('Hash sebelum:')
print(hash_sebelum)

tx_uji['price_per_kg'] = 5500

hash_sesudah = hash_transaction(tx_uji)
print('\nHarga diubah dari Rp6.500/kg menjadi Rp5.500/kg')
print('\nHash sesudah:')
print(hash_sesudah)

print('\nApakah hash sama?', hash_sebelum == hash_sesudah)

if hash_sebelum == hash_sesudah:
    print('Data tidak berubah.')
else:
    print('PERINGATAN: Data transaksi telah berubah!')
```

## Kesimpulan

- SHA-256 menghasilkan fingerprint dari data transaksi pertanian.
- Perubahan kecil pada data menghasilkan hash berbeda.
- `previous_hash` menghubungkan Block 1 dengan Genesis Block.
- Ini adalah demonstrasi blockchain sederhana, bukan jaringan blockchain terdistribusi lengkap.
