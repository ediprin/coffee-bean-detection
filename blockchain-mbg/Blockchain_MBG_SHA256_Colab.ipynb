# Blockchain Sederhana untuk Data Transaksi MBG

Notebook ini mendemonstrasikan alur sederhana:

**Data transaksi MBG → SHA-256 → hash transaksi → block → previous hash**

Tujuannya adalah memahami integritas data dengan teknik kriptografi tanpa membuat sistem blockchain yang rumit.

```python
import hashlib
import json

def sha256(data: str) -> str:
    return hashlib.sha256(data.encode('utf-8')).hexdigest()

def hash_transaction(transaction: dict) -> str:
    transaction_string = json.dumps(
        transaction,
        sort_keys=True,
        separators=(',', ':')
    )
    return sha256(transaction_string)

def hash_block(block: dict) -> str:
    block_string = json.dumps(
        block,
        sort_keys=True,
        separators=(',', ':')
    )
    return sha256(block_string)

print('Fungsi hashing siap digunakan.')
```

## 1. Data Transaksi MBG

Contoh transaksi bahan pangan dari supplier ke SPPG dan pengiriman meal batch ke sekolah.

```python
tx1 = {
    'transaction_id': 'TX001',
    'sender': 'Supplier-A',
    'receiver': 'SPPG-01',
    'item': 'Ayam Batch C08',
    'quantity': 100,
    'unit': 'kg',
    'timestamp': '2026-09-09T08:00:00'
}

tx2 = {
    'transaction_id': 'TX002',
    'sender': 'Supplier-B',
    'receiver': 'SPPG-01',
    'item': 'Beras Batch R02',
    'quantity': 200,
    'unit': 'kg',
    'timestamp': '2026-09-09T08:30:00'
}

tx3 = {
    'transaction_id': 'TX003',
    'sender': 'SPPG-01',
    'receiver': 'SDN-01',
    'item': 'Meal Batch M10',
    'quantity': 500,
    'unit': 'porsi',
    'timestamp': '2026-09-09T11:00:00'
}

transactions = [tx1, tx2, tx3]

for tx in transactions:
    print(tx)
```

## 2. Menghitung Hash SHA-256 Setiap Transaksi

```python
transaction_hashes = []

for tx in transactions:
    tx_hash = hash_transaction(tx)
    transaction_hashes.append(tx_hash)
    print('ID   :', tx['transaction_id'])
    print('Hash :', tx_hash)
    print('-' * 80)
```

## 3. Membuat Genesis Block

```python
genesis_block = {
    'index': 0,
    'previous_hash': '0',
    'timestamp': '2026-09-09T00:00:00',
    'transactions': []
}

genesis_hash = hash_block(genesis_block)

print('Genesis Block:')
print(json.dumps(genesis_block, indent=2))
print('\nGenesis Hash:')
print(genesis_hash)
```

## 4. Membuat Block 1

```python
block1 = {
    'index': 1,
    'previous_hash': genesis_hash,
    'timestamp': '2026-09-09T12:00:00',
    'transaction_hashes': transaction_hashes
}

block1_hash = hash_block(block1)

print('Block 1:')
print(json.dumps(block1, indent=2))
print('\nPrevious Hash :', block1['previous_hash'])
print('Block 1 Hash  :', block1_hash)
```

## 5. Uji Perubahan Data

Kita ubah jumlah ayam dari 100 kg menjadi 101 kg dan bandingkan hash.

```python
tx_uji = tx1.copy()

hash_sebelum = hash_transaction(tx_uji)
print('Hash sebelum:')
print(hash_sebelum)

tx_uji['quantity'] = 101

hash_sesudah = hash_transaction(tx_uji)
print('\nQuantity diubah dari 100 kg menjadi 101 kg')
print('\nHash sesudah:')
print(hash_sesudah)

print('\nApakah hash sama?', hash_sebelum == hash_sesudah)

if hash_sebelum == hash_sesudah:
    print('Data tidak berubah.')
else:
    print('PERINGATAN: Data transaksi telah berubah!')
```

## Kesimpulan

- SHA-256 menghasilkan hash 256-bit yang biasanya ditampilkan sebagai 64 karakter heksadesimal.
- Perubahan kecil pada transaksi menghasilkan hash yang berbeda.
- `previous_hash` menghubungkan Block 1 dengan Genesis Block.
- Notebook ini adalah demonstrasi sederhana, bukan blockchain terdistribusi lengkap.
